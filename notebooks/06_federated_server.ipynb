```python
"""
federated_server.py
Federated Learning Server - Aggregates client model updates
"""

import flwr as fl
from typing import List, Tuple, Dict, Optional
import numpy as np

class SaveModelStrategy(fl.server.strategy.FedAvg):
    """Custom strategy to save aggregated model and track metrics"""
    
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.round_metrics = []
    
    def aggregate_fit(
        self,
        server_round: int,
        results: List[Tuple[fl.server.client_proxy.ClientProxy, fl.common.FitRes]],
        failures: List[BaseException],
    ):
        """Aggregate model updates from clients"""
        
        # Call parent aggregate_fit
        aggregated_parameters, aggregated_metrics = super().aggregate_fit(
            server_round, results, failures
        )
        
        if aggregated_parameters is not None:
            print(f"\n[Server] Round {server_round} aggregation complete")
            print(f"[Server] Received updates from {len(results)} clients")
            
            # Track metrics
            if aggregated_metrics:
                self.round_metrics.append({
                    'round': server_round,
                    'metrics': aggregated_metrics
                })
        
        return aggregated_parameters, aggregated_metrics
    
    def aggregate_evaluate(
        self,
        server_round: int,
        results: List[Tuple[fl.server.client_proxy.ClientProxy, fl.common.EvaluateRes]],
        failures: List[BaseException],
    ):
        """Aggregate evaluation metrics from clients"""
        
        if not results:
            return None, {}
        
        # Call parent aggregate_evaluate
        loss_aggregated, metrics_aggregated = super().aggregate_evaluate(
            server_round, results, failures
        )
        
        # Print aggregated metrics
        if metrics_aggregated:
            print(f"\n[Server] Round {server_round} Evaluation:")
            print(f"[Server]   Aggregated Loss: {loss_aggregated:.4f}")
            if 'precision' in metrics_aggregated:
                print(f"[Server]   Precision@5: {metrics_aggregated['precision']:.4f}")
            if 'recall' in metrics_aggregated:
                print(f"[Server]   Recall@5: {metrics_aggregated['recall']:.4f}")
        
        return loss_aggregated, metrics_aggregated


def start_server(n_rounds: int = 10, n_clients: int = 10):
    """Start Flower server"""
    
    print("="*60)
    print("FEDERATED POI RECOMMENDATION - SERVER")
    print("="*60)
    print(f"\nConfiguration:")
    print(f"  - Number of rounds: {n_rounds}")
    print(f"  - Minimum clients: {n_clients}")
    print(f"  - Server address: localhost:8080")
    print("\nWaiting for clients to connect...")
    print("="*60)
    
    # Define strategy
    strategy = SaveModelStrategy(
        fraction_fit=1.0,  # Use all available clients for training
        fraction_evaluate=1.0,  # Use all available clients for evaluation
        min_fit_clients=n_clients,  # Minimum clients needed for training
        min_evaluate_clients=n_clients,  # Minimum clients needed for evaluation
        min_available_clients=n_clients,  # Minimum clients that must connect
    )
    
    # Start server
    fl.server.start_server(
        server_address="localhost:8080",
        config=fl.server.ServerConfig(num_rounds=n_rounds),
        strategy=strategy,
    )
    
    print("\n" + "="*60)
    print("SERVER COMPLETED")
    print("="*60)


if __name__ == "__main__":
    # Start server with 10 rounds and 10 clients
    start_server(n_rounds=10, n_clients=10)
```
